# 🥊 Competitor Table Builder
## Automated Multi-Dimensional Competitive Intelligence Matrix Generator

**Pack ID:** `pack_0004` · **Category:** `consulting` · **Difficulty:** `intermediate` · **Version:** `1.1.0`

---
Transforms structured competitor data into standardized comparative matrices in both CSV and JSON formats.


```python
import json
import time
from pathlib import Path
import pandas as pd

INPUT_PATH = Path('data/competitors.json')
OUT_CSV = Path('outputs/competitor_table.csv')
OUT_JSON = Path('outputs/competitor_table.json')
OUT_CSV.parent.mkdir(parents=True, exist_ok=True)

with open(INPUT_PATH, 'r', encoding='utf-8') as f:
    data = json.load(f)

items = data.get('competitors', [])
print(f'Loaded {len(items)} competitor profiles.')
```

```python
# Normalize and Compute Relative Feature Score
table_rows = []
for c in items:
    score = (
        (1 if c.get('multi_agent_support') else 0) +
        (1 if c.get('guardrails_included') else 0) +
        (1 if c.get('local_mcp_support') else 0) +
        (1 if c.get('open_source') else 0)
    )
    row = {
        'name': c['name'],
        'tier': c['tier'],
        'price_per_month': c['pricing_monthly_starting'],
        'multi_agent': 'Yes' if c.get('multi_agent_support') else 'No',
        'guardrails': 'Yes' if c.get('guardrails_included') else 'No',
        'mcp_protocol': 'Yes' if c.get('local_mcp_support') else 'No',
        'open_source': 'Yes' if c.get('open_source') else 'No',
        'deployments': ', '.join(c.get('deployment_modes', [])),
        'feature_score': f'{score}/4'
    }
    table_rows.append(row)

df_comp = pd.DataFrame(table_rows)
df_comp.to_csv(OUT_CSV, index=False, encoding='utf-8')
print(f'✅ Written: {OUT_CSV}')

with open(OUT_JSON, 'w', encoding='utf-8') as f:
    json.dump({'industry': data.get('industry'), 'comparison': table_rows}, f, indent=2)
print(f'✅ Written: {OUT_JSON}')

print('\n📋 Comparison Matrix Preview:')
print(df_comp[['name', 'price_per_month', 'feature_score', 'mcp_protocol', 'open_source']].to_string(index=False))
```

```python
# Verification
assert OUT_CSV.exists() and OUT_CSV.stat().st_size > 0
assert OUT_JSON.exists() and OUT_JSON.stat().st_size > 0
print('✅ Pack 0004 execution completed successfully.')
```
